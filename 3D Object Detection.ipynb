```python
import cv2
import mediapipe as mp

cap = cv2.VideoCapture(0)

def rescale_frame(frame, present=75):
    width = int(frame.shape[1] * present / 100)
    height = int(frame.shape[0] * present / 100)
    dim = (width, height)
    return cv2.resize(frame, dim, interpolation=cv2.INTER_AREA)

try:
    with mp.solutions.objectron.Objectron(static_image_mode=False,
                                           max_num_objects=1,
                                           min_detection_confidence=0.5,
                                           min_tracking_confidence=0.5,
                                           model_name="Cup") as objectron:
        while cap.isOpened():
            ret, img = cap.read()

            if not ret:
                break

            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            img.flags.writeable = False

            try:
                result = objectron.process(img)
            except RuntimeError as e:
                print(f"An error occurred: {e}")
                continue

            img.flags.writeable = True
            img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)

            if result.detected_objects:
                print("Object Detected")
                for detected_object in result.detected_objects:
                    mp.solutions.drawing_utils.draw_landmarks(img, detected_object.landmarks_2d,
                                                              mp.solutions.objectron.BOX_CONNECTIONS)
                    mp.solutions.drawing_utils.draw_axis(img, detected_object.rotation,
                                                         detected_object.translation)

            cv2.imshow("3D Object Tracking", rescale_frame(img, present=150))
            if cv2.waitKey(0) == ord("q"):
                break

finally:
    cap.release()
    cv2.destroyAllWindows()
```
